```python
#1. 1부터 키보드로 입력받은 수까지의 합을 출력
inputData = int(input('입력 : '))
inputSum = 0
for e in range(1, inputData + 1) :
  inputSum = inputSum + e

print(inputSum)
```

```python
#2. 키보드로 입력받은 단수의 구구단을 출력
gugudan = int(input('입력 : '))
for e in range(1, 10) :
  print(f'{gugudan} * {e} = {gugudan * e}')
```

```python
#3. 아래 제시된 리스트의 요소 중 가장 큰 수를 출력
list1 = [3, 1, 7, 10, 5]
print(max(list1))
print(min(list1))
```

```python
#4. 아래에 제시된 두 리스트의 모든 요소의 합을 출력
list2 = [1, 3, 5]
list3 = [10, 15, 20]
print(sum(list2) + sum(list3))
print(sum(list2 + list3))
```

```python
#5. 아래의 데이터에서 't'의 갯수를 구하시오.
data = ['mother', 'father', 'jentleman']

tCnt = 0
for word in data :
  for c in word :
    if c == 't' :
      tCnt = tCnt + 1
      
print('t의 갯수 : ', tCnt)
```

```python
# 5번 문제 힌트>.<
# 리스트에서 특정 데이터의 존재 유무를 확인 => 자바스크립트에서는 .includes()
imsi_data = [1, 2, 3]

# 위의 리스트에 3이 포함되어 있으면 
if 3 in imsi_data : 
  print(1)
```
