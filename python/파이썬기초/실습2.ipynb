```python
#3. 아래 딕셔너리 데이터가 매개변수로 주어지면 모든 점수의 합을 리턴하는 함수 선언 및 호출
scores = {
  'kor' : 70,
  'math' : 80,
  'eng' : 90
}

# 함수 선언
#def getSum(scores) :
  #return 70+ 80 + 90

#def totalScore(dic) :
  #return dic['kor'] + dic['math'] + dic['eng']

def getTotalScore(dic) :
  #총합을 저장할 변수
  sumData = 0
  #매개변수로 전달된 모든 점수 수만큼 반복
  for key in dic : 
    sumData = sumData + dic[key]
  return sumData
```

```python
# 3번 문체의 함수를 호출
getTotalScore(scores)
```

```python
'''
4번
매개변수로 전달된 딕셔너리에 주어진 key와 value를 추가한 딕셔너리를 리턴하는 함수를 선언 및 호출
첫 번째 매개변수 : 딕셔너리
두 번째 매개변수 : 추가할 key
세 번째 매개변수 : 추가할 value 
'''

# 4번 함수 선언
def getNewDic(dic, newKey, newValue) :
  dic[newKey] = newValue
  return dic
```

```python
# 4번 함수 호출
person = {
  'name' : 'kim',
  'age' : 20
}

getNewDic(person, 'addr', '울산')
```

```python
'''
5번
매개변수로 문자열 리스트가 전달되면 리스트의 문자열과 문자열의 길이를 딕셔너리로 리턴하는 함수 선언 및 호출

ex) 매개변수 : ['apple', 'banana', 'orange']
result) {
  'apple' : 5,
  'banana' : 6,
  'orange' : 6
}
'''

# 파이썬은 문자열도 리스트로 취급!

# 함수 선언
def getDicFromList(strList) : 
  result = {}
  
  for e in strList : 
    result[e] = len(e)

  return result
```

```python
# 5번 함수 호출
getDicFromList(['a', 'bb', 'ccc', 'dd'])
```
