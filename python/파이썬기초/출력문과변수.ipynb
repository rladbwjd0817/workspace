### 파이썬의 출력문

```python
# 파이썬 주석(코드영역이라서)
# 파이썬은 명령어 끝에 세미콜론 안 붙임!
print(1)
print('gjaowgj')
print("aawskfjow")
print(10-2)
print('fseo' + 'fwofj')
#print('fseo' + 2) 문자열과 숫자 더하기는 오류 남!
```

```python
# 연산자
print(3 + 5)
print(3 * 5)
print(11 / 5) # 2.2 => 그냥 나누기
print(11 // 5) # 2 => 몫만 구하면 '//'
print(11 % 5) # '%' => 나머지
```

### 파이썬의 변수 사용

```python
# 파이썬 변수는 내부적으로만 자료형이 존재
# 변수 선언 시 자료형을 작성하지 않음.

# 이름을 저장할 수 있는 변수 선언 및 초기화
myName = 'yujeong'
print(myName)

myAge = 30
myHeight = 157.9

#type(변수) : 매개변수로 전달된 데이터의 자료형을 확인할 수 있음!
print(type(myName))
print(type(myAge))
print(type(myHeight))
```

```python
# 문자열과 변수를 함께 출력하는 방법
addr = '울산시'
score = 90

# str() : 매개변수로 전달된 숫자를 문자열로 반환
# int() : 매개변수로 전달된 문자열을 정수로 반환

# '사는 곳은 울산시이며, 제 점수는 90점 입니다.'
print('사는 곳은 ' + addr + '이며, 제 점수는 ' + str(score) + '점 입니다.' )

# 문자넣을 땐 '' , 문자열 + 변수 ``
print(f'내가 사는 곳은 {addr}이며, 제 점수는 {score}점 입니다.')
```
