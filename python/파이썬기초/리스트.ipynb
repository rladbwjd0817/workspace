### 리스트 

```python
# 리스트(다수의 데이터를 index 번호로 관리하는 자료형)
# 변수명에 list는 사용 금지!
list1 = [] # 빈 리스트 생성
list2 = [1, 3.5, 'hey yu!'] # 서로 다른 자료형 저장 가능

print(list1)
print(list2)
print(list2[1])
```

```python
# 리스트에 데이터 추가하기
list1 = [1, 2, 3] # 값이 변경되는 코드, 같은 줄에서는 안됨

# 1. append() : 자바스크립트의 push()와 동일한 기능(리스트의 가장 마지막 요소에 데이터 추가)
list1.append(4) #[1, 2, 3, 4]

# 2. insert(위치, 추가할 데이터) : 원하는 index에 데이터 추가(원하는 위치에 넣을 수 있음)
list1.insert(2, 5) #[1, 2, 5, 3, 4]

print(list1)
```

```python
# 리스트의 데이터 삭제

# del 명령어 사용
del list1[1] #list1의 첫번째 요소를 지워라
print(list1)

# 주의사항
# del list1 # list1이라는 변수자체를 삭제
# print(list1) 위 코드의 실행으로 변수 자체가 삭제되기 때문에 오류발생!
# 실행한 코드는 전부 기억하고 있음 컴퓨터가 똑똑행
```

```python
# 리스트 데이터 수정
list2 = [1, 3.3, 'python']

# 1번요소를 'hello'로 변경
list2[1] = 'hello'

print(list2)
```

```python
# 리스트 합치기
list1 = [1, 2, 3]
list2 = [4, 5, 6]

#[1, 2, 3, 4, 5, 6]
list3 = list1 + list2
print(list3)

list4 = list1 * 3 #[1, 2, 3, 1, 2, 3, 1, 2, 3]
print(list4)
```

```python
# 리스트의 슬라이싱 연산 (중요!) : 리스트의 특정 부분만 추출
# [시작 index : 마지막 index(마지막 index는 포함 X)]
list1 = [1, 2, 3, 4, 5]
print(list1[1:4])

#list1 배열에서 [2,3] 데이터만 갖는 새로운 리스트를 생성
list2 = list1[1:3]
print(list2)

list3 = [1,2,3,4,5]
print(list3[2:]) # 2번 index부터 끝까지
print(list3[:3]) # 처음부터 3번 index 전까지
print(list3[:]) # 전부 출력
print(list3[2:-1]) # 음수면 오른쪽 index부터 -1, -2로 증가
```

```python
# 파이썬에서는 문자열을 리스트로 취급!!
myName = 'hong'
print(myName[1])

addrList = [
  '스타벅스 삼산지점 052-1111-2222',
  '스타벅스 병영지점 052-2222-3333',
  '스타벅스 복산점 052-3333-4444'
]

# 스타벅스 매장 정보에서 연락처 정보만 추출
print(addrList[2][-13:])
print(addrList[1][-13:])
```
